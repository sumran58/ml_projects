```python
import pandas as pd
import numpy as np
import difflib
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

```

```python
df = pd.read_csv("/content/movies.csv", engine="python", on_bad_lines='skip')
```

```python
#selecting the relevant features for movie recommendation
selected_features = ['genres','keywords','tagline','cast','director']
print(selected_features)
```

```python
#replacing the null values with null strings
for i in selected_features:
  df[i]=df[i].fillna('')
```

```python
#combinig all the 5 selected features
combined_features=df['genres']+''+df['keywords']+''+df['tagline']+''+df['cast']+''+df['director']
```

```python
print(combined_features)
```

```python
#converting the text data to numerical values
fe=TfidfVectorizer()
```

```python
feature_vectors=fe.fit_transform(combined_features)
```

```python
similarity=cosine_similarity(feature_vectors)
```

```python
movie_name=input("enter your favourite movie name:")
```

```python
list_of_all_titles=df['title'].tolist()
```

```python
find_close_match=difflib.get_close_matches(movie_name,list_of_all_titles)
print(find_close_match)
```

```python
close_match=find_close_match[0]
print(close_match)
```

```python
index_of_movie=df[df.title==close_match]['index'].values[0]
```

```python
print(index_of_movie)
```

```python
similarity_score=list(enumerate(similarity[index_of_movie]))
print(similarity_score)
```

```python
len(similarity_score)
```

```python
sorted_similar_movies=sorted(similarity_score,key=lambda x:x[1],reverse=True)
print(sorted_similar_movies)
```

```python
#printing the names of thr similar movies based on the index
print('Movie suggested for you :\n')
i=1
for movie in sorted_similar_movies:
  index=movie[0]
  title_from_index=df[df.index==index]['title'].values[0]
  if(i<30):
    print(i,'.',title_from_index)
    i+=1
```
